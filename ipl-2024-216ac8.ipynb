# IPL 2024 Data Analysis

## Project Overview
This project analyzes IPL 2024 player statistics to find insights about 
top batsmen, bowlers, and all-rounders using Python, Pandas and Matplotlib.

## Libraries Used
- Pandas — Data loading and analysis
- Matplotlib — Data visualization and graphs

## Key Findings
- Top run scorer in IPL 2024 is the player with highest bars in the chart
- Players with high batting average tend to score more runs
- Bowlers with low economy rate take more wickets

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/kaggle/input/ipl-2024-player-lifetime-dataset/cricket_data.csv')
data.head()
```

```python
top_batsmen = data.sort_values(by='Runs_Scored', ascending=False).head(10)
top_batsmen[['Player_Name', 'Runs_Scored']]
```

```python
print(data['Runs_Scored'].dtype)
print(data['Runs_Scored'].head(10))
```

```python
data['Runs_Scored'] = pd.to_numeric(data['Runs_Scored'], errors='coerce')

top_batsmen = data.sort_values(by='Runs_Scored', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_batsmen['Player_Name'], top_batsmen['Runs_Scored'], color='skyblue')
plt.xlabel('Player Name')
plt.ylabel('Runs Scored')
plt.title('Top 10 IPL Batsmen by Runs Scored')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
data['Wickets_Taken'] = pd.to_numeric(data['Wickets_Taken'], errors='coerce')

top_bowlers = data.sort_values(by='Wickets_Taken', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_bowlers['Player_Name'], top_bowlers['Wickets_Taken'], color='orange')
plt.xlabel('Player Name')
plt.ylabel('Wickets Taken')
plt.title('Top 10 IPL Bowlers by Wickets Taken')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
data['Sixes'] = pd.to_numeric(data['Sixes'], errors='coerce')

top_six_hitters = data.sort_values(by='Sixes', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_six_hitters['Player_Name'], top_six_hitters['Sixes'], color='green')
plt.xlabel('Player Name')
plt.ylabel('Sixes')
plt.title('Top 10 IPL Six Hitters')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
data['Batting_Average'] = pd.to_numeric(data['Batting_Average'], errors='coerce')

top_average = data.sort_values(by='Batting_Average', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.barh(top_average['Player_Name'], top_average['Batting_Average'], color='purple')
plt.xlabel('Batting Average')
plt.ylabel('Player Name')
plt.title('Top 10 Players by Batting Average')
plt.show()
```

```python
data['Batting_Strike_Rate'] = pd.to_numeric(data['Batting_Strike_Rate'], errors='coerce')

top_sr = data.sort_values(by='Batting_Strike_Rate', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_sr['Player_Name'], top_sr['Batting_Strike_Rate'], color='red')
plt.xlabel('Player Name')
plt.ylabel('Strike Rate')
plt.title('Top 10 Players by Batting Strike Rate')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(data['Batting_Average'], data['Runs_Scored'], color='blue', alpha=0.5)
plt.xlabel('Batting Average')
plt.ylabel('Runs Scored')
plt.title('Batting Average vs Runs Scored')
plt.show()
```

```python
data['Economy_Rate'] = pd.to_numeric(data['Economy_Rate'], errors='coerce')

plt.figure(figsize=(10,6))
plt.scatter(data['Economy_Rate'], data['Wickets_Taken'], color='orange', alpha=0.5)
plt.xlabel('Economy Rate')
plt.ylabel('Wickets Taken')
plt.title('Economy Rate vs Wickets Taken')
plt.show()
```

```python
top5_batsmen = data.sort_values(by='Runs_Scored', ascending=False).head(5)

plt.figure(figsize=(8,8))
plt.pie(top5_batsmen['Runs_Scored'], 
        labels=top5_batsmen['Player_Name'], 
        autopct='%1.1f%%',
        colors=['skyblue','orange','green','red','purple'])
plt.title('Top 5 Batsmen - Runs Share')
plt.show()
```
